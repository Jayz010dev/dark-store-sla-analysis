```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder

df_model = df[['City','Company','Distance_Km','Delivery_Time_Min','SLA_Breach']].copy()

le_city = LabelEncoder()
le_company = LabelEncoder()
df_model['City_enc'] = le_city.fit_transform(df_model['City'])
df_model['Company_enc'] = le_company.fit_transform(df_model['Company'])

X = df_model[['City_enc','Company_enc','Distance_Km']]
y = df_model['SLA_Breach']

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

importance = pd.Series(model.feature_importances_, index=X.columns).sort_values(ascending=False)
importance
```
